```python
import pandas as pd
```

```python
etfs = pd.read_excel('data/etf_base_assets.xlsx')
```

```python
etfs
```

```python
# Analyser les colonnes importantes pour comprendre les unités
print("=== STRUCTURE DES DONNÉES ETF ===")
print(f"Nombre d'ETF: {len(etfs)}")
print(f"Nombre de colonnes: {len(etfs.columns)}")
print("\n=== COLONNES DISPONIBLES ===")
print(etfs.columns.tolist())
```

```python
# Analyser spécifiquement les colonnes problématiques
print("\n=== ANALYSE DES FRAIS DE GESTION ===")
if 'FRAIS DE GESTION' in etfs.columns:
    frais_sample = etfs['FRAIS DE GESTION'].dropna().head(10)
    print(f"Échantillon des frais de gestion: {frais_sample.tolist()}")
    print(f"Min: {etfs['FRAIS DE GESTION'].min()}")
    print(f"Max: {etfs['FRAIS DE GESTION'].max()}")
    print(f"Moyenne: {etfs['FRAIS DE GESTION'].mean():.4f}")

print("\n=== ANALYSE DES ENCOURS SOUS GESTION ===")
if 'ENCOURS SOUS GESTION (EUR)' in etfs.columns:
    encours_sample = etfs['ENCOURS SOUS GESTION (EUR)'].dropna().head(10)
    print(f"Échantillon des encours (EUR): {encours_sample.tolist()}")
    print(f"Min: {etfs['ENCOURS SOUS GESTION (EUR)'].min()}")
    print(f"Max: {etfs['ENCOURS SOUS GESTION (EUR)'].max()}")
    print(f"Moyenne: {etfs['ENCOURS SOUS GESTION (EUR)'].mean():.0f}")

print("\n=== ANALYSE DES COLONNES BOOLÉENNES ===")
for col in ['PEA', 'ASSVIE', 'HEDGED']:
    if col in etfs.columns:
        print(f"\n{col}:")
        print(f"  Type: {etfs[col].dtype}")
        print(f"  Valeurs uniques: {etfs[col].unique()}")
        print(f"  Nombre de True: {(etfs[col] == True).sum()}")
        print(f"  Nombre de 'Y': {(etfs[col] == 'Y').sum()}")
        print(f"  Nombre de 1: {(etfs[col] == 1).sum()}")
        print(f"  Nombre de valeurs non-nulles: {etfs[col].notna().sum()}")
```
